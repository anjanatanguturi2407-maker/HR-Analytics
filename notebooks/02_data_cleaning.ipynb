# HR Analytics — Data Cleaning

## Objective
The objective of this notebook is to clean and prepare the HR Analytics dataset for exploratory data analysis (EDA) and employee attrition analysis.

## Data Cleaning Steps

1. **Remove Duplicate Records:** Identify and remove exact duplicate rows to avoid repeated records in the dataset.

2. **Standardize Categorical Values:** Fix inconsistent capitalization, extra spaces, and category labels in columns such as Gender, Department, Attrition, Overtime, JobSatisfaction, PerformanceRating, and EducationLevel.

3. **Clean and Convert Data Types:** Convert numeric columns such as Age, MonthlyIncome, and YearsAtCompany into appropriate numeric formats.

4. **Handle Invalid and Suspicious Values:** Investigate unusual values, including negative income, unrealistic ages, and negative years at the company, before deciding how to handle them.

5. **Handle Missing Values:** Analyze missing data and apply suitable treatment based on the nature of each column.

6. **Validate the Cleaned Dataset:** Check data types, missing values, duplicate records, category consistency, and suspicious values after cleaning.

7. **Save the Cleaned Dataset:** Export the final cleaned dataset for use in the next stage of the project.

## Expected Outcome

A clean, consistent, and analysis-ready HR dataset that can be used to explore employee attrition patterns and investigate factors associated with employee turnover.

# Data Cleaning Summary

## Cleaning Operations Performed

- Removed 30 duplicate rows from the original dataset.
- Standardized categorical values in Gender, Department, Attrition, Overtime, EducationLevel, JobSatisfaction, and PerformanceRating.
- Converted Age, MonthlyIncome, and YearsAtCompany into numeric data types.
- Handled suspicious age values by marking ages below 18 and above 60 as missing.
- Removed invalid negative values from MonthlyIncome by marking them as missing.
- Converted inconsistent YearsAtCompany values such as "five", "3 yrs", and "forty" into numeric values.
- Replaced missing values in categorical columns with "Unknown".
- Filled missing Age values using the median age.
- Filled missing MonthlyIncome values using the median income.
- Filled missing YearsAtCompany values using the median number of years.
- Preserved missing Attrition labels because the actual employee outcomes are unknown.
- Standardized Department labels and consolidated inconsistent department names.
- Verified that EmployeeID contains no missing or duplicate identifiers.

## Missing Value Treatment

- Age: Missing values filled using the median (35 years).
- MonthlyIncome: Missing values filled using the median (26,484).
- YearsAtCompany: Missing values filled using the median (7 years).
- Categorical columns: Missing values replaced with "Unknown".
- Attrition: 22 missing values retained for further analysis.

## Final Dataset Summary

- Total Rows: 1,500
- Total Columns: 12
- Duplicate Rows Remaining: 0
- Duplicate EmployeeIDs: 0
- Total Missing Values: 22
- Missing Values Remaining: Attrition only

## Data Quality Notes

- Six records contain a MonthlyIncome value of 999999. These values were retained because their validity has not been confirmed.
- YearsAtCompany contains a maximum value of 45 years, which requires further contextual verification.
- Missing Attrition labels will not be guessed or imputed.
