```python
print("Job satisfaction distribution")
print(df["JobSatisfaction"].value_counts(dropna=False))
```

```python
# Explanation: This code checks whether JobSatisfaction contains only the four expected categories: Low, Medium, High, and Very High. It ignores capitalization and removes extra spaces. The ~ selects values that don't match, while value_counts(dropna=False) counts them, including missing values. The output shows 20 missing values and no other unexpected categories.
```

```python
print(df.loc[
          ~df["JobSatisfaction"].str.strip().str.fullmatch(
              r"(?i)low|medium|high|very high", na=False
          ),
          "JobSatisfaction"
      ].value_counts(dropna=False))
```

```python
print(df.loc[
          ~df["Attrition"].str.strip().str.fullmatch(
              r"(?i)yes|no", na=False
          ),
          "Attrition"
      ].value_counts(dropna=False))
```
